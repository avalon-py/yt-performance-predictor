```python
import pandas as pd

df = pd.read_csv("data/videos.csv")

def capitalized_letter_count(title):
    return sum(1 for c in str(title) if c.isupper())

df["title_capitalized_letter_count"] = df["title"].apply(capitalized_letter_count)
df["title_capitalized_letter_ratio"] = df.apply(
    lambda row: row["title_capitalized_letter_count"] / len(str(row["title"]))
    if len(str(row["title"])) > 0 else 0.0,
    axis=1
)

df.to_csv("data/videos.csv", index=False)
print(df[["title", "title_capitalized_letter_count", "title_capitalized_letter_ratio"]].head())
```
